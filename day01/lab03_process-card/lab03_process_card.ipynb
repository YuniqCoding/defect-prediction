## 실습 3: 공정 카드
- 목표: 공정 하나를 골라 하는 일, 측정값, 이상 신호를 카드 한 장으로 정리

### 공정 카드 - 식각 (Etch)

**1. 하는 일**

웨이퍼 위 박막에서 패턴 밖 부분을 깎아내 회로 모양을 남긴다.

**2. 측정하는 값**

| 항목 | 단위 | 뜻 |
|---|---|---|
| 식각률 | Å/min (nm/min) | 1분에 깎이는 두께 |
| 챔버 압력 | mTorr | 챔버 안 기압 |
| RF 파워 | W | 플라즈마를 만드는 전력 |
| 가스 유량 | sccm | 넣어주는 반응 가스의 양 |
| 척 온도 | °C | 웨이퍼가 놓인 판의 온도 |

결과 쪽 지표로는 CD(선폭, nm), 식각 균일도(%), 선택비(비율, 단위 없음)도 함께 본다.

**3. 정상 범위를 벗어나면**

| 값 | 벗어났을 때 |
|---|---|
| 식각률 | 높으면 과식각 - 아래층까지 깎임 / 낮으면 미식각 - 남아서 회로가 안 끊김 |
| 챔버 압력 | 낮으면 측벽이 파이고, 높으면 옆으로도 깎여 선폭이 굵어짐 |
| RF 파워 | 높으면 아래층 손상·프로파일 틀어짐 / 낮으면 안 깎이거나 잔류물 남음 |
| 가스 유량 | 흔들리면 식각률과 선택비가 같이 흔들리고, 웨이퍼 안에서 편차 발생 |
| 척 온도 | 편차가 크면 웨이퍼 중앙과 가장자리의 깎인 정도가 달라짐 (균일도 저하) |

**4. 확실하지 않거나 현장·장비마다 다른 부분** ⚠️

| 항목 | 주의 |
|---|---|
| 정상 범위 숫자 | 레시피·막질(옥사이드/폴리/메탈)·장비마다 달라서 일반적인 수치를 적을 수 없음. 현장 스펙 확인 필요 |
| 식각률 | 실시간 측정이 아니라 전후 두께 차이로 계산하는 값. 장비 로그에 직접 안 찍힐 수 있음 |
| 척 온도 | "챔버온도"가 척 온도인지 챔버 벽 온도인지 장비마다 다르게 기록됨 |
| 압력 단위 | mTorr / Pa / mbar 중 무엇을 쓰는지 라인마다 다름 |
| 식각 방식 | 위 내용은 건식(플라즈마) 기준. 습식이면 측정 항목 자체가 달라짐 |

> 현장 데이터를 확인한 게 아니라 일반적인 공정 지식으로 정리한 카드다. 실제 데이터의 컬럼과 대조해볼 것.


### 파고들기 - 가스 유량

**어떤 단위로 재나**

| 단위 | 읽는 법 | 메모 |
|---|---|---|
| sccm | standard cubic centimeter per minute | 현장에서 가장 많이 쓰는 단위 |
| slm | standard liter per minute | sccm의 1000배. 유량이 큰 가스에 씀 |

'standard(표준 상태)'는 정해진 온도·압력 기준으로 환산한 양이라는 뜻이다. 즉 실제 부피가 아니라 **기체의 양(몰 수)** 을 재는 것에 가깝다.
⚠️ 확실하지 않음: 그 '표준 상태'를 0°C로 잡는지 20°C/25°C로 잡는지는 MFC 제조사마다 다르다고 알고 있다. 확인 필요.

**현장에서 왜 중요하게 보나**

| 이유 | 내용 |
|---|---|
| 반응의 재료 | 깎는 일을 실제로 하는 건 가스다. 양이 바뀌면 식각률이 바로 따라 바뀐다 |
| 가스 비율 | 여러 가스를 섞어 쓰는데, 그 **비율**이 선택비와 측벽 모양을 정한다. 한 가스만 흔들려도 비율이 깨진다 |
| 압력과 연결 | 유량이 변하면 챔버 압력도 같이 움직인다. 두 값을 따로 보면 원인을 놓친다 |
| 이상 감지 | 설정값(setpoint)과 실제값이 벌어지면 MFC 고장이나 라인 막힘 신호로 본다 |

**장비·회사마다 다른 부분** ⚠️

| 항목 | 다른 점 |
|---|---|
| 기록되는 값 | 설정값만 남기는 곳, 실제 측정값도 남기는 곳이 갈린다. 데이터에 둘 중 뭐가 들어 있는지 확인해야 함 |
| 컬럼 이름 | `Gas1_Flow`, `MFC1`, `Ar_flow` 등 명명 규칙이 회사마다 다름 |
| 쓰는 가스 종류 | 막질에 따라 다름 (예: 옥사이드에 플루오린계, 메탈에 염소계). ⚠️ 확실하지 않음 - 레시피는 회사 기밀이라 일반화하기 어렵다 |
| 단위 표기 | sccm / slm 혼용. 로그에 단위가 안 적혀 있으면 값의 자릿수로 추정해야 할 수 있다 |
| 허용 오차 | 설정값 대비 몇 %까지 정상으로 보는지는 라인 기준마다 다름. ⚠️ 확실하지 않음 |

> MFC = Mass Flow Controller. 가스 유량을 정해진 값으로 맞춰 흘려주는 부품.
